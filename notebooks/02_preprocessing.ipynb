Clean the raw dataset, engineer features, handle missing data, and create train/val/test splits.

```python
import sys
sys.path.insert(0, '..')

import numpy as np
import pandas as pd
from src.data.preprocessing import load_raw_data, preprocess, build_splits
```

## 1. Run Full Preprocessing Pipeline

```python
train_df, val_df, test_df, scaler, label_encoder = build_splits(
    raw_path='../data/raw/dataset.csv',
    output_dir='../data/processed',
    test_size=0.15,
    val_size=0.15,
    random_state=42,
)
```

## 2. Verify Splits

```python
print('Classes:', label_encoder.classes_)
print()

for name, split_df in [('Train', train_df), ('Val', val_df), ('Test', test_df)]:
    print(f'--- {name} ---')
    print(f'  Shape: {split_df.shape}')
    print(f'  NaN in features: {split_df.drop(columns=["class_label", "h_target", "diameter_target", "h_mask", "diameter_mask"]).isna().sum().sum()}')
    print(f'  Class distribution:')
    print(split_df['class_label'].value_counts().sort_index().to_string())
    print(f'  H coverage: {split_df["h_mask"].mean():.1%}')
    print(f'  Diameter coverage: {split_df["diameter_mask"].mean():.1%}')
    print()
```

```python
# Feature overview after preprocessing
feature_cols = [c for c in train_df.columns if c not in ['class_label', 'h_target', 'diameter_target', 'h_mask', 'diameter_mask']]
print(f'Number of features: {len(feature_cols)}')
print(f'Features: {feature_cols}')
print(f'\nScaled feature stats (train):')
train_df[feature_cols].describe().round(3)
```
